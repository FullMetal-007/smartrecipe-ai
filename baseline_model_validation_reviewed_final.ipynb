# SmartRecipe AI: Baseline Model Validation and Ranking Evaluation

This notebook evaluates the **frozen TF-IDF + combined-embedding retrieval artifacts used by the running app**. It does not retrain the model and does not call Gemini.

Workflow: verify artifacts → load the app's baseline service → retrieve the same nine assessment queries → measure candidate-level constraints and latency → export results → optionally calculate ranking metrics from human judgments.

**Metric validity:** similarity scores are not accuracy probabilities. Precision@K, MRR@K and NDCG@K require human relevance judgments. True Recall@K, F1@K and MAP@K require an independently judged relevant-item pool, not just the same top-K items. No labels or results are fabricated.

**Artifact compatibility:** these pickles were created with scikit-learn 0.24.1. Newer versions can emit an `InconsistentVersionWarning`; use the project's pinned environment where possible and disclose any remaining warning.

```python
baseline_service.load_models()
print("Baseline artifacts loaded.")
print("Recipe rows:", len(baseline_service._sampled_data))
feature_count = (len(baseline_service._vectorizer.get_feature_names_out())
                 if hasattr(baseline_service._vectorizer, "get_feature_names_out")
                 else len(baseline_service._vectorizer.get_feature_names()))
print("TF-IDF feature count:", feature_count)
print("Combined embedding shape:", baseline_service._combined_embeddings.shape)
```

```python
# Load the fixed nine assessment queries.
TEST_CASES_PATH = PROJECT_ROOT / "data" / "assessment_test_cases.json"
test_cases = json.loads(TEST_CASES_PATH.read_text(encoding="utf-8"))
print("Test cases:", len(test_cases))
pd.DataFrame(test_cases)[["test_id", "request", "purpose"]]
```

```python
# Run baseline retrieval only. This does not call Gemini.
# Keep the raw top-five list so the audit shows what retrieval actually returned.
# Cooking time <= 0 is treated as unknown, not automatically compliant.
TOP_K = 5
retrieval_rows = []

for test in test_cases:
    started = time.perf_counter()
    candidates = baseline_service.get_candidates(test["request"], num_candidates=TOP_K)
    elapsed = time.perf_counter() - started
    req = parse_requirements(test["request"])

    # This is the application's hard-constraint filter. We retain the raw
    # candidates below for an audit of baseline retrieval, and report membership
    # in the filter's accepted set separately.
    valid, rejected = filter_candidates_by_constraints(candidates, req)
    valid_keys = {
        (str(c.name), int(c.baseline_rank))
        for c in valid
    }

    for candidate in candidates:
        matched, missing = ingredient_match_lists(candidate, req)
        diet_violations = _diet_violations(candidate, req)

        if not req.max_cooking_minutes:
            time_status = "not_applicable"
            time_compliant = True
        elif candidate.minutes is None or candidate.minutes <= 0:
            time_status = "unknown"
            time_compliant = None
        elif candidate.minutes <= req.max_cooking_minutes:
            time_status = "within_limit"
            time_compliant = True
        else:
            time_status = "over_limit"
            time_compliant = False

        non_time_constraints_ok = (not missing) and (not diet_violations)
        if not non_time_constraints_ok or time_compliant is False:
            hard_ok = 0
        elif time_compliant is None:
            # Unknown cooking time means overall compliance cannot be confirmed.
            hard_ok = np.nan
        else:
            hard_ok = 1

        retrieval_rows.append({
            "test_id": test["test_id"],
            "query": test["request"],
            "rank": candidate.baseline_rank,
            "recipe_name": candidate.name,
            "similarity_score": candidate.baseline_similarity_score,
            "minutes": candidate.minutes,
            "ingredients": ", ".join(candidate.ingredients),
            "tags": ", ".join(candidate.tags),
            "matched_required_ingredients": ", ".join(matched),
            "missing_required_ingredients": ", ".join(missing),
            "diet_violations": "; ".join(diet_violations),
            "time_status": time_status,
            "time_compliant": time_compliant,
            "hard_constraints_passed": hard_ok,
            "accepted_by_app_constraint_filter": (
                (str(candidate.name), int(candidate.baseline_rank)) in valid_keys
            ),
            "retrieval_latency_seconds": round(elapsed, 6),
        })

baseline_results = pd.DataFrame(retrieval_rows)
display(baseline_results.head(15))
```

```python
# Summary of actual retrieval and constraint results.
# The hard-constraint rate uses only candidates whose constraints can be assessed;
# unknown cooking times are not counted as passes or failures.
summary = baseline_results.groupby("test_id").agg(
    candidates_returned=("recipe_name", "count"),
    top_similarity=("similarity_score", "max"),
    confirmed_hard_constraint_passes=("hard_constraints_passed", lambda s: int((s == 1).sum())),
    candidates_with_unknown_time=("time_status", lambda s: int((s == "unknown").sum())),
    time_compliant_candidates=("time_compliant", lambda s: int((s == True).sum())),
    accepted_by_app_filter=("accepted_by_app_constraint_filter", "sum"),
    retrieval_latency_seconds=("retrieval_latency_seconds", "first"),
).reset_index()

assessable = baseline_results[baseline_results["hard_constraints_passed"].notna()]
assessable_rate = (
    100 * assessable["hard_constraints_passed"].mean()
    if not assessable.empty else np.nan
)
summary["assessable_hard_constraint_pass_rate_top5_pct"] = (
    summary["test_id"].map(
        assessable.groupby("test_id")["hard_constraints_passed"].mean() * 100
    ).round(2)
)
display(summary)
print("Assessable candidate hard-constraint pass rate (%):",
      round(assessable_rate, 2) if pd.notna(assessable_rate) else "not available")
print("Mean retrieval latency (s):", round(summary["retrieval_latency_seconds"].mean(), 4))
print("Important: this is a constraint audit, not model accuracy or recommendation quality.")
```

## Ranking metrics from human judgments

The next cell uses `outputs/baseline_relevance_judgments.csv` if present. Precision@5, MRR@5, and binary NDCG@5 can be computed from judged top-five results. **Recall@5 and MAP@5 are not computed from only the same top-five items**, because that would make the denominator incomplete. They require total relevant items from a broader, independently judged pool. Do not fill labels automatically or fabricate them.


```python
# Calculate only metrics supported by the available labels, and reject stale/mismatched labels.
judgment_path = OUTPUT_DIR / "baseline_relevance_judgments.csv"

if not judgment_path.exists():
    print("No human judgment file found. Review baseline_candidates_for_judgment.csv and save labels first.")
else:
    judgments = pd.read_csv(judgment_path)
    required_cols = {"test_id", "rank", "recipe_name", "relevance"}

    if not required_cols.issubset(judgments.columns):
        print("Judgment file is missing required columns:", required_cols - set(judgments.columns))
    elif judgments[list(required_cols)].isna().any().any():
        print("Judgment file has missing values in required columns.")
    elif (judgments["relevance"].astype(str).str.strip() == "").any():
        print("Human relevance labels are incomplete. Fill relevance with 1 (relevant) or 0 (not relevant).")
    else:
        judgments["rank"] = pd.to_numeric(judgments["rank"], errors="raise").astype(int)
        judgments["relevance"] = pd.to_numeric(judgments["relevance"], errors="raise").astype(int)

        if not judgments["relevance"].isin([0, 1]).all():
            raise ValueError("Relevance labels must be binary 0/1.")
        if (judgments["rank"] < 1).any():
            raise ValueError("Candidate ranks must be positive integers.")

        # Reject duplicate labels for the same candidate.
        judgment_keys = ["test_id", "rank", "recipe_name"]
        if judgments.duplicated(judgment_keys).any():
            duplicates = judgments.loc[judgments.duplicated(judgment_keys, keep=False), judgment_keys]
            raise ValueError("Duplicate relevance-judgment rows found:\n" + duplicates.to_string(index=False))

        # Ensure each judged row still matches the current candidate identity.
        current_ids = set(zip(
            baseline_results["test_id"].astype(str),
            baseline_results["rank"].astype(int),
            baseline_results["recipe_name"].astype(str)
        ))
        judged_ids = list(zip(
            judgments["test_id"].astype(str),
            judgments["rank"].astype(int),
            judgments["recipe_name"].astype(str)
        ))
        mismatches = [x for x in judged_ids if x not in current_ids]
        if mismatches:
            raise ValueError(
                f"{len(mismatches)} relevance labels do not match current baseline candidates. "
                f"Re-review/export labels before scoring; first mismatch: {mismatches[0]}"
            )

        # Each query needs a complete set of top-five labels to report P@5/MRR@5/NDCG@5.
        counts = judgments.groupby("test_id").size()
        incomplete = counts[counts != 5]
        if not incomplete.empty:
            raise ValueError(
                "Each evaluated query must have exactly five unique top-five judgments. "
                f"Queries with a count other than five: {incomplete.to_dict()}"
            )

        def _dcg(rel):
            rel = np.asarray(rel, dtype=float)
            return float(np.sum((2**rel - 1) / np.log2(np.arange(2, len(rel) + 2))))

        metric_rows = []
        for test_id, group in judgments.groupby("test_id"):
            group = group.sort_values("rank")
            rel = group["relevance"].tolist()
            p5 = sum(rel) / 5
            first = next((i for i, value in enumerate(rel, 1) if value == 1), None)
            ideal_dcg = _dcg(sorted(rel, reverse=True))
            metric_rows.append({
                "test_id": test_id,
                "Precision@5": p5,
                "MRR@5": 1 / first if first else 0.0,
                "NDCG@5": _dcg(rel) / ideal_dcg if ideal_dcg else np.nan,
                "relevant_in_top5": sum(rel),
                "Recall@5": np.nan,
                "F1@5": np.nan,
                "MAP@5": np.nan
            })

        ranking_metrics = pd.DataFrame(metric_rows)
        display(ranking_metrics.round(3))
        display(
            ranking_metrics[["Precision@5", "MRR@5", "NDCG@5"]]
            .mean().rename("mean_score").to_frame().round(3)
        )
        print(
            "Recall@5/F1@5/MAP@5 remain blank because these labels cover only retrieved "
            "top-five items, not an independently judged relevant pool."
        )
```

## Interpretation and limitations

- This notebook evaluates the **frozen saved retrieval artifacts**, not a newly trained classifier.
- Similarity scores are ranking signals, not accuracy probabilities.
- Relevance metrics depend on human judgments and the evaluation pool.
- The raw top-five candidates are audited before/alongside the application's hard-constraint filter; the `accepted_by_app_constraint_filter` column records the filter's decision.
- A cooking time of zero, negative, or missing is treated as **unknown** when a maximum time is requested. It is not silently counted as compliant.
- Candidate-level hard-constraint pass rate is not the same as user-facing recommendation success rate.
- Allergy safety cannot be certified from ingredient text alone.
- Keep the original EDA/preprocessing/advanced modelling notebooks as model-development lineage. Cite or describe their outputs only if you actually ran and inspected them; the live app's baseline provenance is the saved artifacts validated above.
